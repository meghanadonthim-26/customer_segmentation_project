```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 120

df = pd.DataFrame({
    "CustomerID": [f"C{i:03d}" for i in range(1, n+1)],
    "Age": np.random.randint(18, 66, n),
    "AnnualIncome": np.random.randint(20000, 120001, n),
    "AnnualSpending": np.random.randint(5000, 60001, n),
    "PurchasesPerYear": np.random.randint(1, 31, n),
    "WebsiteVisitsPerMonth": np.random.randint(1, 41, n)
})

df.head()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

features = [
    "Age",
    "AnnualIncome",
    "AnnualSpending",
    "PurchasesPerYear",
    "WebsiteVisitsPerMonth"
]

X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

df["Segment"] = kmeans.fit_predict(X_scaled)

df.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(
    df["AnnualIncome"],
    df["AnnualSpending"],
    c=df["Segment"]
)

plt.xlabel("Annual Income")
plt.ylabel("Annual Spending")
plt.title("Customer Segmentation")

plt.show()
```

```python
print(df.groupby("Segment")[features].mean().round(2))
```
